1-2. 문자열을 숫자로 변환하기

```python
houses = []
excluded_count = 0

for index, row in df_house.iterrows():

    if pd.isna(row[["ID", "size", "year", "price"]]).any():
        print(index, "번 행 제외: 값 누락")
        excluded_count += 1
        continue

    try:
        house = {
            "ID": int(row["ID"]),
            "size": int(row["size"]),
            "year": int(row["year"]),
            "price": float(row["price"])
        }

        houses.append(house)

    except(ValueError, TypeError):
        print(index, "번 행 제외: 숫자 변환 오류")
        excluded_count += 1

print("정상 저장된 주택 수:", len(houses))
print("제외된 주택 수:", excluded_count)
```

```python
df_test = df_house.copy()

df_test.loc[10] = [11, None, 5, 3.0]
df_test.loc[11] = [12, "오류", 5, 3.0]
```

```python
houses = []
excluded_count = 0

for index, row in df_test.iterrows():

    if pd.isna(row[["ID", "size", "year", "price"]]).any():
        print(index, "번 행 제외: 값 누락")
        excluded_count += 1
        continue

    try:
        house = {
            "ID": int(row["ID"]),
            "size": int(row["size"]),
            "year": int(row["year"]),
            "price": float(row["price"])
        }

        houses.append(house)

    except(ValueError, TypeError):
        print(index, "번 행 제외: 숫자 변환 오류")
        excluded_count += 1

print("정상 저장된 주택 수:", len(houses))
print("제외된 주택 수:", excluded_count)
```

1-3. 조건에 맞는 주택 찾기

```python
houses_test = houses.copy()

houses_test.append({"ID": 101, "year": 5, "price": 3})
houses_test.append({"ID": 102, "size": 35, "year": 5, "price": None})
houses_test.append({"ID": 103, "size": "오류", "year": 5, "price": 3})
houses_test.append({"ID": 104, "size": 35, "year": 5, "price": 3})

print(len(houses))
print(len(houses_test))
```

```python
selected = []

for house in houses_test:

    if "size" not in house or "price" not in house:
        print(house["ID"], "제외: 키 누락")
        continue

    if pd.isna(house["size"]) or pd.isna(house["price"]):
        print(house["ID"], "제외:값 누락")
        continue

    try:
        size = float(house["size"])
        price = float(house["price"])

    except (ValueError, TypeError):
        print(house["ID"], "제외: 숫자 변환 오류")
        continue

    if size <= 0 or price < 0:
        print(house["ID"], "제외: 유효하지 않은 값")
        continue

    if size >= 30 and price <= 4:
        selected.append(house)

for house in selected:
    print("선택된 주택 ID:", house["ID"])

print("선택된 주택 수:", len(selected))
```

1-4. 함수와 예외 처리

```python
def unit_price(price, size):

    try:
        price = float(price)
        size = float(size)

        if size <= 0:
            print("크기 오류: 크기는 0보다 커야 합니다.")
            return None

        if price < 0:
            print("가격 오류: 가격은 0 이상이어야 합니다.")
            return None

        return price / size

    except ValueError:
        print("값 오류: 숫자로 변환할 수 없는 문자열입니다.")
        return None

    except TypeError:
        print("자료형 오류: 숫자 또는 숫자 형태의 문자열을 입력하세요.")
        return None
```

```python
for house in houses:
    result = unit_price(house["price"], house["size"])

    print("주택 ID:", house["ID"], "크기당 가격:", result)
```

```python
print(unit_price(4, 40))
print(unit_price("4", "40"))
print(unit_price(4, 0))
print(unit_price(-4, 40))
print(unit_price("오류", 40))
print(unit_price(None, 40))
print(unit_price(6, 30))
```

1-5. 목적에 맞는 자료구조 선택하기

```python
# 1. 주택 ID 저장
# 순서대로 저장해야 해서 리스트를 사용함
# set은 순서대로 저장된다는 보장이 없음

house_ids = []

for house in houses:
    house_ids.append(house["ID"])

print("주택 ID:", house_ids)


# 2. 중복되는 year 제거
# 리스트는 같은 값이 여러 번 들어갈 수 있음
# set은 중복된 값을 제거해 주기 때문에 사용함

house_years = set()

for house in houses:
    house_years.add(house["year"])

print("중복 제거된 year:", house_years)


# 3. ID로 주택 가격 찾기
# 리스트는 위치 번호로 값을 찾지만
# 딕셔너리는 ID를 키로 사용해서 가격을 찾을 수 있음

house_prices = {}

for house in houses:
    house_prices[house["ID"]] = house["price"]

print("ID 3번 주택의 가격:", house_prices[3])


# 4. 주택 선택 조건 저장
# 리스트는 저장한 값을 바꿀 수 있음
# 30과 4는 바꿀 필요가 없는 조건이라 튜플을 사용함

conditions = (30, 4)

print("주택 선택 조건:", conditions)
```
